20 - Find the length of last word in a string

    Input: s = "Python is an easy to learn, powerful programming language"
    Output: 8
    Explanation: The last word is "language" with length 8.

```python
s = "Python is an easy to learn, powerful programming language"
words = s.split()
if len(words) > 0:
    last_word = words[-1]
    length_last_word = len(last_word)
    print("The last word is",last_word,"with length",length_last_word)
else:
    print("No words found in the string.")
```
